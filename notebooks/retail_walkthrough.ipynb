# Retail Forecasting Walkthrough

This notebook is a beginner-friendly walkthrough using synthetic data generated for the project.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('../data/raw/retail_sales.csv')
df.head()
```

```python
df['date'] = pd.to_datetime(df['date'])
weekly = df.set_index('date')['sales_qty'].resample('W').mean()
weekly.plot(figsize=(10,4), title='Weekly Average Sales')
plt.show()
```
